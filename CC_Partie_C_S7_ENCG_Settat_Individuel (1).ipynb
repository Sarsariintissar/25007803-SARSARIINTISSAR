 # 1. Régression linéaire

### 1.1 Présentation

La régression linéaire est un algorithme d'apprentissage supervisé qui permet de prédire une valeur numérique à partir d'une ou plusieurs variables.

L'équation générale est : Ŷ = β₀ + β₁X.

- Ŷ : valeur prédite.
- X : variable explicative.
- β₀ : constante.
- β₁ : coefficient de la variable.

### 1.2 Schéma original

Données d'entrée (X et Y) → Apprentissage du modèle → Équation de régression → Prédiction et évaluation.

### 1.3 Contextes et choix

- Commerce : prévoir les ventes selon les dépenses publicitaires.
- Finance : estimer les revenus d'une entreprise selon ses investissements.

On choisit cet algorithme lorsqu'on souhaite prédire une valeur numérique et que la relation entre les variables est approximativement linéaire.

### 1.4 Test de sensibilité et interprétation

Nous avons modifié le paramètre fit_intercept pour observer son effet sur les performances du modèle.

- True : le modèle calcule une constante.
- False : le modèle force la constante à zéro.

Les résultats obtenus sont :

- Avec constante (True) : MAE = 41,55 et R² = 0,485.
- Sans constante (False) : MAE = 156,36 et R² = -3,940.

On remarque que le modèle avec constante donne de meilleurs résultats. Son MAE est plus faible et son R² est positif. En revanche, sans constante, le MAE augmente fortement et le R² devient négatif.

Cela montre que la constante est importante pour ce jeu de données. Nous retenons donc le modèle avec fit_intercept=True, car il offre de meilleures performances de prédiction.

# 2. Ridge

### 2.1 Présentation

Ridge est un algorithme d'apprentissage supervisé qui permet de prédire une valeur numérique. Il ajoute une pénalisation pour limiter la taille des coefficients et éviter le surapprentissage.

**Équation :**

J(β) = (1/n) ||y - Xβ||² + α Σ βⱼ²

- J(β) : fonction à minimiser.
- y : valeurs réelles.
- X : variables explicatives.
- β : coefficients du modèle.
- n : nombre d'observations.
- α : force de la pénalisation.

### 2.2 Schéma original

Données → Standardisation → Régression Ridge → Prédiction → Évaluation.

### 2.3 Contextes et choix

- Finance : prévoir les revenus à partir de plusieurs indicateurs financiers.
- Commerce : prévoir les ventes à partir de plusieurs facteurs commerciaux.

Ridge est choisi lorsque plusieurs variables explicatives sont fortement corrélées, car il permet de stabiliser les coefficients.

**Question spécifique :** lorsque alpha augmente, la pénalisation devient plus forte et les coefficients se rapprochent progressivement de zéro, sans être généralement annulés.

### 2.4 Test de sensibilité et interprétation

Nous avons utilisé le paramètre alpha pour étudier son effet sur le modèle.

Le modèle initial utilise alpha = 12 et donne un MAE de 41,46.

Nous allons comparer trois valeurs d'alpha (1, 12 et 100) afin d'observer l'évolution des coefficients et des performances du modèle.

Nous avons modifié le paramètre **alpha** en utilisant trois valeurs : 1, 12 et 100, afin d'observer son effet sur les coefficients et les performances du modèle.

**Résultats obtenus :**

- Alpha = 1 : MAE = 41,51.
- Alpha = 12 : MAE = 41,46.
- Alpha = 100 : MAE = 41,93.

**Interprétation :**

On remarque que lorsque alpha augmente, la pénalisation devient plus forte et les coefficients ont tendance à se rapprocher de zéro.

Concernant les performances, le modèle donne le meilleur résultat avec alpha = 12, car il présente le MAE le plus faible (41,46).

Lorsque alpha passe à 100, le MAE augmente à 41,93, ce qui indique une légère dégradation des performances.

**Conclusion :** nous retenons alpha = 12, car cette valeur offre le meilleur résultat parmi les trois valeurs testées.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error

for alpha in [1, 12, 100]:
    modele = make_pipeline(
        StandardScaler(),
        Ridge(alpha=alpha)
    )

    modele.fit(Xr_ent, yr_ent)
    predictions = modele.predict(Xr_test)

    ridge = modele.named_steps['ridge']

    print("Alpha :", alpha)
    print("MAE :", round(mean_absolute_error(yr_test, predictions), 2))
    print("Coefficients :", ridge.coef_)
    print("----------------------")
```

## 3. Lasso

### 3.1 Présentation

Lasso est un algorithme d'apprentissage supervisé qui permet de prédire une valeur numérique. Il ajoute une pénalisation L1 qui peut annuler certains coefficients et effectuer une sélection automatique des variables.

**Équation :**

J(β) = (1/n) ||y - Xβ||² + α Σ |βⱼ|

- J(β) : fonction à minimiser.
- y : valeurs réelles.
- X : variables explicatives.
- β : coefficients du modèle.
- n : nombre d'observations.
- α : force de la pénalisation.

### 3.2 Schéma original

Données → Standardisation → Régression Lasso → Sélection des variables → Prédiction → Évaluation.

### 3.3 Contextes et choix

- Finance : identifier les facteurs qui influencent les revenus.
- Marketing : sélectionner les variables qui influencent les ventes.

Lasso est utilisé lorsqu'on souhaite sélectionner les variables les plus importantes et éliminer celles qui sont moins utiles.

### 3.4 Test de sensibilité et interprétation

Le modèle initial donne les résultats suivants :

- Nombre de coefficients non nuls : 8.
- MAE Lasso : 41,37.

Cela signifie que le modèle conserve 8 variables et annule les coefficients des autres variables. L'objectif du test de sensibilité est de comparer trois valeurs d'alpha pour observer l'évolution du nombre de variables annulées et des performances du modèle.
Nous avons modifié le paramètre alpha en utilisant trois valeurs : 0,01 ; 0,1 et 1.

Résultats obtenus :

Alpha = 0,01 : 10 coefficients non nuls, 0 annulé, MAE = 41,54.

Alpha = 0,1 : 9 coefficients non nuls, 1 annulé, MAE = 41,48.

Alpha = 1 : 9 coefficients non nuls, 1 annulé, MAE = 41,40.

Interprétation :

On remarque que lorsque alpha augmente, la pénalisation devient plus forte et certains coefficients peuvent être annulés. Le nombre de variables conservées passe de 10 à 9.

Concernant les performances, le meilleur résultat est obtenu avec alpha = 1, car le MAE est le plus faible (41,40).

Conclusion : Lasso permet de sélectionner automatiquement les variables importantes. Dans notre test, alpha = 1 donne le meilleur MAE parmi les trois valeurs étudiées.


```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error
import numpy as np

for alpha in [0.01, 0.1, 1]:
    modele = make_pipeline(
        StandardScaler(),
        Lasso(alpha=alpha, max_iter=10000)
    )

    modele.fit(Xr_ent, yr_ent)
    predictions = modele.predict(Xr_test)

    lasso = modele.named_steps['lasso']
    non_nuls = np.sum(lasso.coef_ != 0)
    annules = np.sum(lasso.coef_ == 0)

    print("Alpha :", alpha)
    print("Nombre de coefficients non nuls :", non_nuls)
    print("Nombre de coefficients annulés :", annules)
    print("MAE Lasso :", round(mean_absolute_error(yr_test, predictions), 2))
    print("----------------------")
```

## 4. Régression polynomiale

### 4.1 Présentation

La régression polynomiale est un algorithme d'apprentissage supervisé qui permet de modéliser une relation non linéaire entre les variables explicatives et la variable à prédire.

**Équation :**

\[
\hat{Y}=\beta_0+\beta_1X+\beta_2X^2+\cdots+\beta_dX^d
\]

- Ŷ : valeur prédite.
- X : variable explicative.
- β₀ : constante.
- β₁, β₂ : coefficients du modèle.
- d : degré du polynôme.

### 4.2 Schéma original

Données → Standardisation → Transformation polynomiale → Régression Ridge → Prédiction → Évaluation.

### 4.3 Contextes et choix

- Économie : étudier l'évolution des prix en fonction de la demande.
- Commerce : analyser les ventes lorsque leur évolution n'est pas linéaire.

La régression polynomiale est utilisée lorsque la relation entre les variables est courbe et ne peut pas être correctement représentée par une droite.
### 4.4 Test de sensibilité et interprétation

Nous avons modifié le paramètre **degree** en comparant deux valeurs : 1 et 2.

**Résultats obtenus :**

- Degré 1 :
  - R² entraînement = 0,502.
  - R² test = 0,490.

- Degré 2 :
  - R² entraînement = 0,557.
  - R² test = 0,501.

**Interprétation :**

On remarque que le passage du degré 1 au degré 2 améliore le R² sur les données d'entraînement, qui passe de 0,502 à 0,557. Il améliore également le R² sur les données de test, qui passe de 0,490 à 0,501.

Cela signifie que le degré 2 améliore légèrement les performances du modèle sur les nouvelles données. L'amélioration ne concerne donc pas uniquement l'entraînement.

**Conclusion :** nous retenons le degré 2, car il donne un meilleur résultat sur les données de test. Cependant, l'amélioration reste faible.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score

for degree in [1, 2]:
    modele = make_pipeline(
        StandardScaler(),
        PolynomialFeatures(degree=degree, include_bias=False),
        Ridge(alpha=100.)
    )

    modele.fit(Xr_ent, yr_ent)

    pred_ent = modele.predict(Xr_ent)
    pred_test = modele.predict(Xr_test)

    print("Degré :", degree)
    print("R² entraînement :", round(r2_score(yr_ent, pred_ent), 3))
    print("R² test :", round(r2_score(yr_test, pred_test), 3))
    print("----------------------")
```

## 5. Arbre de régression

### 5.1 Présentation

L'arbre de régression est un algorithme d'apprentissage supervisé qui permet de prédire une valeur numérique en divisant les données en plusieurs groupes à partir de règles de décision.

**Principe mathématique :**

L'arbre cherche à minimiser l'erreur quadratique moyenne dans les feuilles :

\[
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

- MSE : erreur quadratique moyenne.
- n : nombre d'observations.
- yᵢ : valeur réelle.
- ŷᵢ : valeur prédite.

### 5.2 Schéma original

Données → Division selon des critères → Création des branches → Feuilles → Prédiction → Évaluation.

### 5.3 Contextes et choix

- Immobilier : prédire le prix d'un logement selon sa surface et son emplacement.
- Commerce : prévoir les ventes selon le prix et la demande.

L'arbre de régression est utile lorsque les relations entre les variables sont complexes et qu'on souhaite obtenir un modèle facile à interpréter.
### 5.4 Test de sensibilité et interprétation

Nous avons modifié le paramètre **max_depth** en testant quatre valeurs : 2, 3, 5 et 10.

**Résultats obtenus :**

- Max_depth = 2 : MAE entraînement = 48,00 ; MAE test = 47,33.
- Max_depth = 3 : MAE entraînement = 44,66 ; MAE test = 44,57.
- Max_depth = 5 : MAE entraînement = 36,71 ; MAE test = 47,01.
- Max_depth = 10 : MAE entraînement = 33,63 ; MAE test = 50,27.

**Interprétation :**

On remarque que lorsque la profondeur augmente, l'erreur sur les données d'entraînement diminue. Cependant, l'erreur sur les données de test augmente à partir de la profondeur 3.

Avec une profondeur faible (2), l'arbre est trop simple et présente un biais élevé : il n'apprend pas suffisamment les relations dans les données.

## 7. Gradient Boosting — régression

### 7.1 Présentation

Le Gradient Boosting est un algorithme d'apprentissage supervisé qui combine plusieurs arbres de décision construits successivement. Chaque nouvel arbre cherche à corriger les erreurs des arbres précédents afin d'améliorer les prédictions.

**Équation :**

\[
F_M(x)=F_0(x)+\sum_{m=1}^{M}\eta h_m(x)
\]

- Fₘ(x) : prédiction finale du modèle.
- F₀(x) : prédiction initiale.
- M : nombre d'itérations.
- η : taux d'apprentissage.
- hₘ(x) : correction apportée par le nouvel arbre.

### 7.2 Schéma original

Données → Modèle initial → Calcul des erreurs → Nouvel arbre correcteur → Répétition → Prédiction finale.

### 7.3 Contextes et choix

- Finance : prévoir les risques ou les revenus.
- Commerce : prévoir les ventes et la demande.

Le Gradient Boosting est utilisé pour améliorer progressivement les prédictions en corrigeant les erreurs des modèles précédents.
### 7.4 Test de sensibilité et interprétation

Nous avons modifié le paramètre **max_iter** en testant quatre valeurs : 50, 100, 200 et 300 itérations.

**Résultats obtenus :**

- 50 itérations : MAE entraînement = 28,68 ; MAE test = 43,13 ; temps = 0,168 s.
- 100 itérations : MAE entraînement = 21,42 ; MAE test = 44,18 ; temps = 0,305 s.
- 200 itérations : MAE entraînement = 13,50 ; MAE test = 45,81 ; temps = 0,347 s.
- 300 itérations : MAE entraînement = 9,00 ; MAE test = 46,75 ; temps = 0,518 s.

**Interprétation :**

On remarque que lorsque le nombre d'itérations augmente, l'erreur d'entraînement diminue, ce qui signifie que le modèle apprend de mieux en mieux les données d'entraînement.

Cependant, l'erreur de test augmente progressivement et le temps d'exécution devient plus important. Cela montre un risque de surajustement : le modèle apprend trop précisément les données d'entraînement et généralise moins bien sur les nouvelles données.

**Conclusion :** nous retenons max_iter = 50, car cette valeur donne le meilleur résultat sur les données de test (MAE = 43,13), avec le temps d'exécution le plus faible (0,168 seconde).

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele = HistGradientBoostingRegressor(max_iter=100, max_leaf_nodes=9, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE boosting :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
import time

for iterations in [50, 100, 200, 300]:
    modele = HistGradientBoostingRegressor(
        max_iter=iterations,
        max_leaf_nodes=9,
        random_state=GRAINE
    )

    debut = time.perf_counter()
    modele.fit(Xr_ent, yr_ent)
    duree = time.perf_counter() - debut

    pred_ent = modele.predict(Xr_ent)
    pred_test = modele.predict(Xr_test)

    print("Max_iter :", iterations)
    print("MAE entraînement :", round(mean_absolute_error(yr_ent, pred_ent), 2))
    print("MAE test :", round(mean_absolute_error(yr_test, pred_test), 2))
    print("Temps d'exécution :", round(duree, 3), "secondes")
    print("----------------------")
```

# 17. K-means

## 1. Présentation

K-means est un algorithme d'apprentissage non supervisé utilisé pour regrouper des observations en plusieurs groupes appelés clusters.

Il cherche à former des groupes homogènes en rapprochant les observations qui se ressemblent.

**Équation :**

\[
J=\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2
\]

- \(J\) : inertie totale.
- \(K\) : nombre de clusters.
- \(C_k\) : groupe numéro \(k\).
- \(x_i\) : observation.
- \(\mu_k\) : centre du groupe (centroïde).

**Résultat initial :** avec 3 clusters, l'inertie est de 1277,93 et le score de silhouette est de 0,285.

## 2. Schéma original

Données → Choix du nombre de clusters → Attribution des observations aux groupes → Calcul des centroïdes → Répétition → Clusters finaux.

## 3. Contextes et choix

K-means est utilisé dans plusieurs domaines :
- La segmentation des clients.
- Le regroupement des produits.
- L'analyse des comportements des consommateurs.
- La segmentation des marchés.

Il est choisi parce qu'il permet de regrouper automatiquement des observations similaires sans connaître les catégories à l'avance.

Deux indicateurs permettent d'évaluer le résultat :
- **Inertie :** mesure la dispersion des observations autour de leur centroïde. Plus elle est faible, mieux les observations sont regroupées.
- **Silhouette :** mesure la qualité de séparation des clusters. Plus elle est proche de 1, meilleure est la séparation.
## 4. Test de sensibilité et interprétation

Nous avons testé trois valeurs du nombre de clusters : k = 2, k = 3 et k = 4.

**Résultats obtenus :**

- k = 2 : inertie = 1658,76 ; silhouette = 0,259.
- k = 3 : inertie = 1277,93 ; silhouette = 0,285.
- k = 4 : inertie = 1175,35 ; silhouette = 0,259.

**Interprétation :**

On remarque que l'inertie diminue lorsque le nombre de clusters augmente. Cela signifie que les observations sont plus proches de leurs centres.

Cependant, le score de silhouette est plus élevé avec k = 3 (0,285). Cela indique que les groupes sont mieux séparés avec trois clusters qu'avec deux ou quatre.

Même si k = 4 donne une inertie plus faible, sa silhouette diminue à 0,259. Il ne représente donc pas le meilleur regroupement selon cet indicateur.

**Conclusion :** nous retenons k = 3, car il offre le meilleur score de silhouette (0,285) et un bon équilibre entre la cohésion et la séparation des groupes.

```python
from sklearn.cluster import KMeans
modele = KMeans(n_clusters=3, n_init=15, random_state=GRAINE).fit(X_clu)
print('Inertie K-means :', round(modele.inertia_, 2))
print('Silhouette :', round(silhouette_score(X_clu, modele.labels_), 3))
```

# 19. DBSCAN

## 1. Présentation

DBSCAN est un algorithme d'apprentissage non supervisé qui permet de regrouper les observations selon leur proximité. Il détecte les groupes de points denses et identifie les observations isolées comme du bruit.

**Équation :**

\[
N_{\varepsilon}(x)=\{y\in X \mid d(x,y)\leq\varepsilon\}
\]

- \(N_{\varepsilon}(x)\) : voisinage du point \(x\).
- \(x\) : observation étudiée.
- \(y\) : autre observation.
- \(d(x,y)\) : distance entre deux observations.
- \(\varepsilon\) (eps) : distance maximale pour considérer deux points comme voisins.

**Résultat initial :** avec eps = 2,5 et min_samples = 4, DBSCAN détecte 1 groupe et 23 points considérés comme bruit.

## 2. Schéma original

Données → Calcul des distances → Identification des points voisins → Formation des groupes denses → Détection des points isolés (bruit).

## 3. Contextes et choix

DBSCAN est utilisé dans plusieurs domaines :
- La détection des anomalies.
- La segmentation des clients.
- L'analyse des données géographiques.
- La détection des comportements inhabituels.

Il est choisi parce qu'il peut identifier des groupes de formes différentes et détecter automatiquement les observations isolées, sans fixer à l'avance le nombre de clusters.
## 4. Test de sensibilité et interprétation

Nous avons testé trois valeurs du paramètre `eps` : 1,5 ; 2,5 et 3,5, en gardant `min_samples = 4`.

**Résultats obtenus :**

- eps = 1,5 : 2 groupes et 169 points bruit.
- eps = 2,5 : 1 groupe et 23 points bruit.
- eps = 3,5 : 1 groupe et 6 points bruit.

**Interprétation :**

On remarque que le paramètre `eps` influence directement le nombre de points considérés comme bruit.

Lorsque eps = 1,5, le voisinage est petit. Plusieurs observations ne trouvent pas suffisamment de voisins et sont donc considérées comme du bruit (169 points).

Lorsque eps augmente à 2,5 puis à 3,5, le voisinage devient plus grand. Le nombre de points bruit diminue alors de 169 à 23, puis à 6.

**Conclusion :** nous retenons eps = 2,5 comme compromis, car il permet de détecter un groupe avec seulement 23 points bruit. Avec eps = 3,5, le nombre de points bruit est encore plus faible, mais un voisinage trop grand peut regrouper des observations qui ne sont pas réellement similaires.

```python
from sklearn.cluster import DBSCAN
etiquettes = DBSCAN(eps=2.5, min_samples=4).fit_predict(X_clu)
print('Groupes détectés hors bruit :', len(set(etiquettes) - {-1}))
print('Points considérés comme bruit :', np.sum(etiquettes == -1))
```

# 20. Mélanges gaussiens

## 1. Présentation

Les mélanges gaussiens (Gaussian Mixture Models) sont des algorithmes d'apprentissage non supervisé qui permettent de regrouper les observations en plusieurs groupes appelés clusters.

Ils supposent que les données sont générées par plusieurs distributions gaussiennes. Chaque observation peut appartenir à plusieurs groupes avec des probabilités différentes.

**Équation :**

\[
p(x)=\sum_{k=1}^{K}\pi_k\mathcal{N}(x\mid\mu_k,\Sigma_k)
\]

- \(p(x)\) : probabilité de l'observation.
- \(K\) : nombre de composantes gaussiennes.
- \(\pi_k\) : poids de la composante \(k\).
- \(\mu_k\) : moyenne de la composante.
- \(\Sigma_k\) : matrice de covariance.
- \(\mathcal{N}\) : distribution normale (gaussienne).

**Résultat initial :** avec 3 composantes et une covariance de type diag, les poids obtenus sont 0,352 ; 0,287 et 0,361. Le score de silhouette est de 0,281.

## 2. Schéma original

Données → Choix du nombre de composantes → Estimation des distributions gaussiennes → Calcul des probabilités d'appartenance → Attribution aux groupes.

## 3. Contextes et choix

Les mélanges gaussiens sont utilisés dans plusieurs domaines :
- La segmentation des clients.
- L'analyse des comportements des consommateurs.
- La reconnaissance des formes.
- La détection des anomalies.

Ils sont choisis parce qu'ils permettent une classification probabiliste : chaque observation possède une probabilité d'appartenance à chaque groupe.

Deux types de covariance sont comparés :

- **Full :** chaque groupe possède une matrice de covariance complète, qui permet de représenter les relations entre les variables.
- **Diag :** utilise uniquement les variances individuelles et suppose que les covariances entre variables sont nulles au sein de chaque groupe.
## 4. Test de sensibilité et interprétation

Nous avons comparé deux types de covariance, full et diag, en gardant le nombre de composantes fixé à 3.

**Résultats obtenus :**

- Covariance full : poids = [0,364 ; 0,286 ; 0,350] ; silhouette = 0,285.
- Covariance diag : poids = [0,352 ; 0,287 ; 0,361] ; silhouette = 0,281.

**Interprétation :**

On remarque que la méthode full obtient un score de silhouette légèrement supérieur (0,285) à celui de diag (0,281).

Cela signifie que full produit des groupes légèrement mieux séparés. Cette méthode prend en compte les relations entre les variables grâce à une matrice de covariance complète.

En revanche, diag suppose que les variables ne sont pas corrélées à l'intérieur de chaque groupe. Elle est plus simple, mais peut être moins adaptée lorsque les variables présentent des relations entre elles.

**Conclusion :** nous retenons le type full, car il donne le meilleur score de silhouette (0,285). Il représente donc le meilleur choix pour nos données dans cette expérience.

```python
from sklearn.mixture import GaussianMixture
modele = GaussianMixture(n_components=3, covariance_type='diag', random_state=GRAINE).fit(X_clu)
etiquettes = modele.predict(X_clu)
print('Poids du mélange :', np.round(modele.weights_, 3))
print('Silhouette :', round(silhouette_score(X_clu, etiquettes), 3) if len(set(etiquettes)) > 1 else 'non définie')
```

## 21. ACP — Analyse en composantes principales

### 1. Présentation

L'ACP est une méthode d'apprentissage non supervisé qui permet de réduire le nombre de variables tout en conservant le maximum d'informations possible.

Elle transforme plusieurs variables initiales en nouvelles variables appelées **composantes principales**.

**Équation :**

\[
Z=XW
\]

- \(X\) : matrice des données initiales.
- \(W\) : matrice des vecteurs propres.
- \(Z\) : nouvelles données projetées.
- Les composantes principales sont des combinaisons des variables initiales.

### 2. Schéma original

Variables initiales → ACP → Composantes principales → Visualisation en 2D ou 3D.

### 3. Contextes et choix

L'ACP est utilisée pour :
- Réduire le nombre de variables.
- Simplifier l'analyse des données.
- Visualiser des données complexes.
- Limiter la redondance entre variables.

Elle est particulièrement utile lorsque le jeu de données contient beaucoup de variables. Le nombre de composantes est choisi selon le pourcentage de variance que l'on souhaite conserver.
### 4. Test de sensibilité et interprétation

Nous avons testé le nombre de composantes principales afin de déterminer combien sont nécessaires pour conserver au moins 80 % de la variance.

**Résultats obtenus :**

- 3 composantes : 66,5 %.
- 4 composantes : 73,6 %.
- 5 composantes : 80,2 %.
- 6 composantes : 85,1 %.

**Interprétation :**

On remarque que le pourcentage de variance expliquée augmente avec le nombre de composantes.

Avec 3 composantes, nous conservons seulement 66,5 % de l'information. En revanche, avec 5 composantes, nous atteignons 80,2 %, ce qui dépasse l'objectif de 80 %.

**Conclusion :** nous retenons donc 5 composantes principales, car elles permettent de conserver 80,2 % de l'information tout en réduisant le nombre de variables initiales.

```python
from sklearn.decomposition import PCA
modele = PCA(n_components=3, random_state=GRAINE).fit(X_clu)
projection = modele.transform(X_clu)
print('Variance expliquée par composante :', np.round(modele.explained_variance_ratio_, 3))
plt.scatter(projection[:, 0], projection[:, 1], s=18)
plt.title('Wine — projection ACP sans étiquettes'); plt.show()
```

```python

from sklearn.decomposition import PCA
import numpy as np

modele = PCA(random_state=GRAINE).fit(X_clu)

variance_cumulee = np.cumsum(modele.explained_variance_ratio_)

nombre = np.argmax(variance_cumulee >= 0.80) + 1

print("Variance expliquée cumulée :", np.round(variance_cumulee, 3))
print("Nombre de composantes nécessaires :", nombre)
print("Variance obtenue :", round(variance_cumulee[nombre-1], 3))
```

## 22. t-SNE

### 1. Présentation

t-SNE (t-distributed Stochastic Neighbor Embedding) est une méthode d'apprentissage non supervisé utilisée pour réduire la dimension des données et les représenter en 2D ou en 3D.

Elle permet de visualiser les groupes et les relations entre les observations en conservant principalement les proximités entre les points.

**Principe mathématique :**

t-SNE cherche à rapprocher dans la représentation finale les observations qui sont proches dans les données initiales.

Il minimise une divergence appelée divergence de Kullback-Leibler :

\[
KL(P\|Q)=\sum_{i\ne j}p_{ij}\log\left(\frac{p_{ij}}{q_{ij}}\right)
\]

- \(P\) : probabilités de proximité entre les points dans les données initiales.
- \(Q\) : probabilités de proximité dans la représentation en 2D.
- \(p_{ij}\) : proximité entre les points \(i\) et \(j\) dans les données initiales.
- \(q_{ij}\) : proximité entre ces points dans la représentation finale.

### 2. Schéma original

Données initiales → Calcul des proximités → Réduction en 2D → Visualisation des groupes.

### 3. Contextes et choix

t-SNE est utilisé pour :
- Visualiser des données complexes.
- Identifier des groupes d'observations.
- Explorer les ressemblances entre les individus.
- Représenter des données avec plusieurs variables sur un graphique en 2D.

Il est particulièrement utile pour l'exploration visuelle, mais il ne garantit pas que les distances entre les groupes représentent fidèlement les distances réelles dans les données initiales.

```python
from sklearn.manifold import TSNE
projection = TSNE(n_components=2, perplexity=12, learning_rate='auto', init='pca', random_state=GRAINE).fit_transform(X_clu)
plt.scatter(projection[:, 0], projection[:, 1], s=18)
plt.title('Wine — t-SNE (perplexity=12)'); plt.show()
```

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```
